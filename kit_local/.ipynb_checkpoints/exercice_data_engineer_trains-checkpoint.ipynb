# Exercice technique – Data Engineer : ponctualité des trains

## Contexte
Vous rejoignez l'équipe data d'un opérateur ferroviaire. Les données d'horaires (théoriques et réels) arrivent de plusieurs systèmes, avec une qualité variable. L'équipe **Performance** veut un tableau de bord fiable de la **ponctualité** par ligne et par mois.

## Mission
1. **Explorer et diagnostiquer** la qualité des 4 sources (anomalies, incohérences, doublons, formats).
2. **Nettoyer et transformer** les données en un modèle propre et jointable.
3. **Produire la table finale** `punctuality_by_line_month` avec, par ligne et par mois : nombre de trains programmés, annulés, retard moyen à l'arrivée finale, et **taux de ponctualité** (part des trains circulés arrivés avec moins de 5 min de retard).
4. **Mettre en place 3 contrôles qualité automatisés** (ex. unicité, valeurs nulles, intégrité référentielle).
5. **Bonus** : décrivez en quelques lignes comment industrialiser ce traitement (orchestration, tests, monitoring).

## Sources (dossier `data/`)
| Fichier | Contenu |
|---|---|
| `stations.csv` | référentiel des gares |
| `routes.csv` | lignes et type de train |
| `trips.csv` | circulations (1 train, 1 jour) |
| `stop_times.csv` | horaires théoriques et réels par arrêt (les heures théoriques suivent la convention GTFS) |
| `incidents.json` | incidents déclarés (cause, impact) |

## Règles du jeu
- **PySpark** (session `spark` déjà créée, mode local) ou Spark SQL. pandas/DuckDB restent possibles pour de petits contrôles, mais la transformation principale doit être faite en PySpark.
- L'assistant IA est **autorisé et encouragé**. Nous évaluons la façon dont vous le pilotez et vérifiez ses résultats, pas son absence.
- Documentez vos hypothèses dans des cellules Markdown.
- Durée : **2 h**.

## 0. Setup (à exécuter tel quel)

````
import os, shutil, sys, importlib.util, urllib.request

# 1) Dépendances (déjà présentes si vous utilisez requirements.txt ou une image Docker Spark)
if importlib.util.find_spec("pyspark") is None:
    %pip install -q pyspark==3.5.3
if shutil.which("java") is None:
    raise RuntimeError("Java est introuvable : PySpark en a besoin (Java 11 ou 17 recommandé).")

# 2) Données : lues dans ./data ; téléchargées seulement si absentes ET si une URL est fournie
DATA_BASE_URL = None   # ex. "https://raw.githubusercontent.com/<ORG>/<REPO>/main/data" (usage Colab)
FILES = ["stations.csv", "routes.csv", "trips.csv", "stop_times.csv", "incidents.json"]
os.makedirs("data", exist_ok=True)
for f in FILES:
    if not os.path.exists(f"data/{f}"):
        if DATA_BASE_URL is None:
            raise FileNotFoundError(f"data/{f} manquant : placez le dossier data/ à côté du notebook.")
        urllib.request.urlretrieve(f"{DATA_BASE_URL}/{f}", f"data/{f}")
print("Données prêtes :", sorted(os.listdir("data")))

# 3) Session Spark (mode local)
os.environ.setdefault("PYSPARK_PYTHON", sys.executable)
from pyspark.sql import SparkSession, functions as F, Window
spark = (SparkSession.builder.master("local[*]").appName("exercice-trains")
         .config("spark.sql.session.timeZone", "UTC")
         .config("spark.sql.shuffle.partitions", "8").getOrCreate())
spark.sparkContext.setLogLevel("ERROR")
print("Spark", spark.version)
````

## Assistant IA
Vous pouvez utiliser l'assistant IA fourni avec la commande `%%ai` : écrivez votre question dans une cellule commençant par `%%ai`. Le code proposé apparaît dans une nouvelle cellule : à vous de le relire, l'exécuter et le corriger.

Vos échanges sont enregistrés dans `prompts_log.jsonl` et font partie de l'évaluation, au même titre que votre code.

*Fournisseur par défaut : Gemini (clé gratuite). Voir `AI_PROVIDER` dans la cellule ci-dessous.*

````
import json, re, datetime, os, getpass

AI_PROVIDER = "gemini"   # "gemini" (gratuit avec clé AI Studio) | "ollama" (100 % local, sans clé) | "anthropic" (payant)
MODELS = {"gemini": "gemini-3.5-flash", "ollama": "qwen2.5-coder:7b", "anthropic": "claude-sonnet-4-6"}

_SYSTEM_BASE = ("Tu es un assistant pour un data engineer qui travaille dans un notebook avec PySpark 3.5 (mode local). "
                "La SparkSession `spark` et `F` (pyspark.sql.functions) existent déjà. "
                "Réponds de façon concise, avec le code PySpark (ou Spark SQL) dans un bloc ```python.")
AI_CONTEXT = ""  # rempli après le chargement des données (schémas)
_history = []    # [{"role": "user"|"assistant", "content": str}]
_clients = {}

def _key(name):
    try:
        from google.colab import userdata          # secret Colab
        return userdata.get(name)
    except Exception:
        return os.environ.get(name) or getpass.getpass(f"{name} : ")

def _ask_gemini(system):
    try:
        from google import genai
    except ImportError:
        get_ipython().run_line_magic("pip", "install -q google-genai")
        from google import genai
    if "gemini" not in _clients:
        _clients["gemini"] = genai.Client(api_key=_key("GEMINI_API_KEY"))
    contents = [{"role": "user" if h["role"] == "user" else "model", "parts": [{"text": h["content"]}]} for h in _history]
    r = _clients["gemini"].models.generate_content(model=MODELS["gemini"], contents=contents,
                                                    config={"system_instruction": system})
    return r.text

def _ask_ollama(system):
    import requests   # nécessite Ollama lancé en local : https://ollama.com (ollama pull qwen2.5-coder:7b)
    msgs = [{"role": "system", "content": system}] + _history
    r = requests.post("http://localhost:11434/api/chat", json={"model": MODELS["ollama"], "messages": msgs, "stream": False}, timeout=300)
    r.raise_for_status()
    return r.json()["message"]["content"]

def _ask_anthropic(system):
    try:
        import anthropic
    except ImportError:
        get_ipython().run_line_magic("pip", "install -q anthropic")
        import anthropic
    if "anthropic" not in _clients:
        _clients["anthropic"] = anthropic.Anthropic(api_key=_key("ANTHROPIC_API_KEY"))
    r = _clients["anthropic"].messages.create(model=MODELS["anthropic"], max_tokens=2000, system=system, messages=_history)
    return r.content[0].text

_PROVIDERS = {"gemini": _ask_gemini, "ollama": _ask_ollama, "anthropic": _ask_anthropic}

from IPython.core.magic import register_cell_magic
from IPython import get_ipython

@register_cell_magic
def ai(line, cell):
    """Usage : %%ai  puis votre question (collez schémas / erreurs si besoin)."""
    _history.append({"role": "user", "content": cell})
    txt = _PROVIDERS[AI_PROVIDER](_SYSTEM_BASE + AI_CONTEXT)
    _history.append({"role": "assistant", "content": txt})
    with open("prompts_log.jsonl", "a") as f:
        f.write(json.dumps({"ts": datetime.datetime.now().isoformat(), "provider": AI_PROVIDER, "prompt": cell, "response": txt}, ensure_ascii=False) + "\n")
    print(txt)
    m = re.search(r"```(?:python|sql)?\n(.*?)```", txt, re.S)
    if m:
        try:
            get_ipython().set_next_input(m.group(1))
        except Exception:
            pass
````

## Aperçu rapide des données

````
stations   = spark.read.option("header", True).option("inferSchema", True).csv("data/stations.csv")
routes     = spark.read.option("header", True).csv("data/routes.csv")
trips      = spark.read.option("header", True).csv("data/trips.csv")
stop_times = spark.read.option("header", True).option("inferSchema", True).csv("data/stop_times.csv")
incidents  = spark.read.option("multiLine", True).json("data/incidents.json")

DFS = dict(stations=stations, routes=routes, trips=trips, stop_times=stop_times, incidents=incidents)
for n, df in DFS.items():
    df.createOrReplaceTempView(n)          # utilisable en Spark SQL
    print(f"{n}: {df.count()} lignes")
trips.show(3); stop_times.show(3, truncate=False); incidents.printSchema()

# Contexte transmis automatiquement à l'assistant IA (%%ai)
AI_CONTEXT = "\n\nTables/DataFrames disponibles (vues temporaires Spark SQL du même nom) :\n" + \
    "\n".join(f"- {n}: {df.dtypes}" for n, df in DFS.items())
````
